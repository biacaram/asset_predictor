```python
import pandas as pd
from models.baseline import NaiveBaseline, HistoricalMeanBaseline, LinearRegressionModel
from backtesting.walk_forward import run_backtest, summarize_backtest

df = pd.read_parquet("data/processed/BTC.parquet")
feature_cols = [c for c in df.columns if c not in ["date", "symbol", "target_return"]]

models = {
    "naive": lambda: NaiveBaseline(),
    "historical_mean": lambda: HistoricalMeanBaseline(),
    "linear_regression": lambda: LinearRegressionModel(),
}

results = run_backtest(df, feature_cols, "target_return", models, n_splits=5, min_train_size=100, test_size=20)
print(summarize_backtest(results))
```
